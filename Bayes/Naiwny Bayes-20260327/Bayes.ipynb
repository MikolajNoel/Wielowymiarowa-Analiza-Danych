## Klasyfikacja za pomocą GausianNB

```python
import sys
import numpy
print(sys.executable)
print(numpy.__version__)
```

```python
import pandas as pd
diabetes = pd.read_csv("diabetes.csv")
diabetes.head()
```

```python
diabetes.columns
```

```python
X = diabetes.drop(columns ='Outcome') # dataframe
y = diabetes['Outcome']  # series

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(X_train,y_train)
y_train_pred = model.predict(X_train)

y_test_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

cm = confusion_matrix(y_train,y_train_pred)
cm
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_) 
                          
disp.plot()

plt.show()
```

```python
cm = confusion_matrix(y_test,y_test_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_) 
                          
disp.plot()

plt.show()
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score, precision_score
```

$\text{Confusion Matrix} =\begin{bmatrix}
\text{True Negative} & \text{False Positive} \\
\text{False Negative} & \text{True Positive} 
\end{bmatrix}$

$\text{accuracy}=\frac{TN+TP}{TN+FN+FP+TP}=\frac{\text{well-predicted observations}}{\text{all observations}}$

$\text{precision}=\frac{TP}{FP+TP}=\frac{\text{well-predicted positive}}{\text{all observations predicted as positive}}$

$\text{recall}=\frac{TP}{FN+TP}=\frac{\text{well-predicted positve}}{\text{all observations which are positive}}$

```python
accuracy_score(y_test,y_test_pred)
```

precision = ile kobiet spośród tych, które uznano, że ma cukrzycę ma ją naprawdę

```python
precision_score(y_test,y_test_pred)
```

recall = ile przypadków cukrzycy zostało poprawnie zinedtyfikowane

```python
recall_score(y_test,y_test_pred)
```

## Detektor spamu

Kod i dane z projektu Julii Krysiak i Wiktorii Poniatowskiej

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.style
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
#importujemy dane
spam_data=pd.read_csv('spam.csv', encoding='latin-1')
spam_data.head()
```

```python
#pozbywamy sie niepotrzebnych kolumn
spam_data=spam_data.drop(["Unnamed: 2","Unnamed: 3","Unnamed: 4"],axis=1)
spam_data=spam_data.rename(columns={"v1":"label","v2":"text"})
spam_data.head()
```

```python
#jaki jest rozklad naszych danych
print('Label distribution (%):', round(spam_data.label.value_counts(normalize=True),3)*100)
plt.figure(figsize=(8,2))
ax=sns.countplot(data=spam_data, y='label')
```

```python
#oznaczamy spam jako 1 oraz ham jako 0
spam_data.loc[:,'label']=spam_data.label.map({'ham':0, 'spam':1})
spam_data['label'] = spam_data['label'].astype(int)
spam_data.head()
```

```python
#dzielimy dane na czesc uczaca i testowa
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split

count = CountVectorizer()
text = count.fit_transform(spam_data['text'])
#Train & test split
x_train, x_test, y_train, y_test = train_test_split(text, spam_data['label'], test_size=0.30, random_state=100)
```

```python
display('X-Train :', x_train.shape)
display('X-Test :',x_test.shape)
display('Y-Train :',y_train.shape)
display('Y-Test :',y_test.shape)
```

```python
%%time
# Multinomial Naive Bayes
from sklearn.naive_bayes import MultinomialNB

#trenujemy model
multinomial_nb_model = MultinomialNB()
multinomial_nb_model.fit(x_train, y_train)  

prediction = multinomial_nb_model.predict(x_test)

print("Multinomial NB")
print("Accuracy score: {}". format(accuracy_score(y_test, prediction)) )
print("Precision score: {}". format(precision_score(y_test, prediction)) )
print("Recall score: {}". format(recall_score(y_test, prediction)))
print("F1 score: {}". format(f1_score(y_test, prediction)))
```
