```python
import pandas as pd
```

# Dane Treningowe


```python
kredyt_trening = pd.read_csv('kredyt.csv')

kredyt_trening.head()
```

```python
kredyt_trening.info()
```

```python
kredyt_trening['Gender'] = kredyt_trening['Gender'].map({'Female': 1, 'Male': 0})
kredyt_trening['Married'] = kredyt_trening['Married'].map({'Yes': 1, 'No': 0})
kredyt_trening['Dependents'] = kredyt_trening['Dependents'].map({'3+': 3, '2': 2, '1': 1, '0': 0})
kredyt_trening['Education'] = kredyt_trening['Education'].map({'Graduate': 1, 'Not Graduate': 0})
kredyt_trening['Self_Employed'] = kredyt_trening['Self_Employed'].map({'Yes': 1, 'No': 0})
kredyt_trening['Loan_Status'] = kredyt_trening['Loan_Status'].map({'Y': 1, 'N': 0})
kredyt_trening = pd.get_dummies(kredyt_trening, columns=['Property_Area'], prefix= 'Area', prefix_sep='_', dtype=int)
kredyt_trening = kredyt_trening.dropna(how = "any")
kredyt_trening
```

# Dane testowe

```python
kredyt_test = pd.read_csv('kredyt_test.csv')
```

```python
kredyt_test
```

```python
kredyt_test = kredyt_test.drop(columns='Loan_ID')
kredyt_test['Gender'] = kredyt_test['Gender'].map({'Female': 1, 'Male': 0})
kredyt_test['Married'] = kredyt_test['Married'].map({'Yes': 1, 'No': 0})
kredyt_test['Dependents'] = kredyt_test['Dependents'].map({'3+': 3, '2': 2, '1': 1, '0': 0})
kredyt_test['Education'] = kredyt_test['Education'].map({'Graduate': 1, 'Not Graduate': 0})
kredyt_test['Self_Employed'] = kredyt_test['Self_Employed'].map({'Yes': 1, 'No': 0})
kredyt_test = pd.get_dummies(kredyt_test, columns=['Property_Area'], prefix= 'Area', prefix_sep='_', dtype=int)
kredyt_test = kredyt_test.dropna(how = "any")
kredyt_test
```

# Dopasowywanie modelu

```python
X = kredyt_trening.drop(columns=['Loan_Status', 'Loan_ID'])
y = kredyt_trening['Loan_Status']
```

```python
param_grid = {
    'max_depth': [3, 4, 5, 6, 8, 10],
    'criterion': ['gini', 'entropy'],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn import tree

base_tree = tree.DecisionTreeClassifier(random_state=42)
grid_search = GridSearchCV(estimator=base_tree, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X, y)
```

```python
grid_search.best_params_
```

```python
grid_search.best_score_
```

```python
best_model = grid_search.best_estimator_
```

```python
y_pred = best_model.predict(kredyt_test)
y_pred = pd.Series(y_pred, name='Loan_Status')
y_pred.to_csv('k.csv', index=False)
```

# Wizualizacje 


```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(20,10))
plot_tree(best_model, feature_names=X.columns, class_names=['Odmowa', 'Zgoda'], filled=True, rounded=True, )
plt.title("Wizualizacja drzewa decyzyjnego")
plt.show()
```

```python
import seaborn as sns

importances = best_model.feature_importances_
feature_names = X.columns
feat_importances = pd.Series(importances, index=feature_names).sort_values(ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(x=feat_importances.values, y=feat_importances.index, palette="flare")
plt.title("Ranking ważności cech w modelu")
plt.xlabel("Poziom istotności")
plt.show()
```
